```python
import pandas as pd

df = pd.read_csv("/content/diabetes.csv")

rows_before = len(df)

df_clean = df[(df['Glucose'].notna()) & (df['Glucose'] != 0)]

rows_after = len(df_clean)

rows_removed = rows_before - rows_after

print("Rows removed:", rows_removed)
print("Rows remaining:", rows_after)
```

i was able to see the numbers of rows removed and obes remaining

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/diabetes.csv")

plt.figure(figsize=(8, 5))
plt.hist(df['Glucose'], bins=20)
plt.title("Distribution of Glucose Levels")
plt.xlabel("Glucose Level")
plt.ylabel("Number of Patients")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/diabetes.csv")

plt.figure(figsize=(8, 5))
plt.scatter(df['Glucose'], df['BMI'])

plt.title("Glucose vs BMI")
plt.xlabel("Glucose")
plt.ylabel("BMI")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/diabetes.csv")

outcome_counts = df['Outcome'].value_counts().sort_index()

plt.figure(figsize=(6, 4))
plt.bar(outcome_counts.index, outcome_counts.values)

plt.title("Number of Patients by Outcome")
plt.xlabel("Outcome")
plt.ylabel("Number of Patients")
plt.xticks([0, 1], ["No Diabetes", "Diabetes"])

plt.show()
```

Bar Chart

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/diabetes.csv")

plt.figure(figsize=(6, 4))
plt.boxplot(df['Glucose'])

plt.title("Box Plot of Glucose Levels")
plt.ylabel("Glucose")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/diabetes.csv")

sns.set_style("whitegrid")  # Other options: darkgrid, white, dark, ticks
```

```python
sns.scatterplot(data=df, x='Glucose', y='BMI', hue='Outcome')

plt.title("Glucose vs BMI")
plt.show()
```

```python
sns.countplot(data=df, x='Outcome')

plt.title("Number of Patients by Outcome")
plt.show()
```

```python
sns.boxplot(data=df, x='Outcome', y='BMI')

plt.title("BMI by Outcome")
plt.show()
```
